데이터 그룹화와 집계

전처리한 데이터들을 분석 목적에 맞게 그룹별로 묶어서 요약하기.

-지역구별 분석 (지역구) : 거래건수, 평균, 중앙값.

-월별 분석 (계약연월) : 거래건수, 평균 거래금액, 중앙값.

-연도별 분석 (계약연도) : 거래건수, 평균 거래금액, 중앙값.

-면적별 분석 (면적구간) : 거래건수, 평균 거래금액, 중앙값.

-전용면적(㎡)당 거래 분석 (면적구간) : ㎡당 거래금액 평균, 중앙값, 건수.

-층 별 분석 (층) : 거래건수, 평균, 중앙값.

 


```python
#지역구별 평균거래금액, 중앙값, 거래건수 집계
district_status = df.groupby("지역구")["거래금액(만원)"].agg(
    평균거래금액="mean",
    중앙거래금액="median",
    거래건수="count"
).reset_index().sort_values(by="평균거래금액", ascending=False).reset_index(drop=True)
district_status
```

```python
#연도별 평균거래금액, 중앙거래금액, 거래건수 집계
df["계약연도"] = df["계약년월"] // 100
year_status = df.groupby("계약연도")["거래금액(만원)"].agg(
    연별평균="mean",
    연별중앙값="median",
    연별거래건수="count"
)
year_status
```

```python
#월별 평균거래금액, 중앙거래금액, 거래건수 집계.
monthly_status = df.groupby("계약년월")["거래금액(만원)"].agg(
    월별평균="mean",
    월별중앙값="median",
    월별거래건수="count"
)
monthly_status
```

```python
#면적 구간별 평균거래금액, 중앙거래금액, 거래건수 집계
df["면적구간"] = pd.cut(
    df["전용면적(㎡)"],
    bins=[0, 40, 60, 85, 135, float("inf")],
    labels=["40㎡ 미만", "40~60㎡", "60~85㎡", "85~135㎡", "135㎡ 이상"],
    right=False
)
area_status = df.groupby("면적구간", observed=True)["거래금액(만원)"].agg(
    평균거래금액="mean",
    중앙거래금액="median",
    거래건수="count"
).reset_index()
area_status
```

```python
#㎡당 거래금액 분석(각 아파트 1㎡당 거래금액(만원))
df["㎡당거래금액"] = (df["거래금액(만원)"] / df["전용면적(㎡)"]).round(2)
df["㎡당거래금액"]
```

```python
#㎡당 거래금액을 평균값, 중앙값, 거래건수로 집계
area_result = df.groupby("면적구간", observed=True)["㎡당거래금액"].agg(
    **{"평균㎡당거래금액": "mean", "중앙㎡당거래금액": "median", "거래건수": "count"}
).reset_index()
area_result
```

```python
#층별 거래금액 분석(층별구간을 나눠서 분석)
df["층구간"] = pd.cut(
    df["층"],
    bins=[-float("inf"), 0, 1, 2, 3, 4, 5, 10, 19, 29, 39, 49, float("inf")],
    labels=[
        "지하층", "1층", "2층", "3층", "4층", "5층",
        "6~10층", "11~19층", "20~29층",
        "30~39층", "40~49층", "50층 이상"
    ],
    right=True
)
floor_status = df.groupby("층구간", observed=True)["거래금액(만원)"].agg(
    평균거래금액="mean",
    중앙거래금액="median",
    거래건수="count"
).reset_index()
floor_status
```

```python
# MariaDB에 최종 데이터 저장

tables = {
    "apartment_transactions": df,
    "district_stats": district_status,
    "year_stats": year_status,
    "monthly_stats": monthly_status,
    "area_stats": area_status,
    "area_price_sqm_stats": area_result,
    "floor_stats": floor_status
}

for table_name, data in tables.items():
    data.to_sql(
        name=table_name,
        con=engine,
        if_exists="replace",
        index=False,
        chunksize=1000
    )
    print(f"{table_name}: {len(data):,}건 저장 완료")
```
