```python
import pandas as pd
import numpy as np
import glob

# CSV 파일 경로
path = "/Users/chy_mac/Desktop/Seoul_Apartment_Project.py/서울_아파트_*.csv"

files = sorted(glob.glob(path))

# 통합 파일 제외
files = [file for file in files if "통합" not in file]

print("찾은 파일:", len(files))
print(files)

# CSV 파일 합치기
df_list = []

for file in files:
    temp = pd.read_csv(file, encoding="cp949", skiprows=15)
    df_list.append(temp)

df = pd.concat(df_list, ignore_index=True)

print("데이터 크기:", df.shape)
display(df.head())
```

```python
# 서울 25개 구별 평균 거래금액 계산
```

```python
district_status = df.groupby("구")["거래금액(만원)"].mean().round(2).sort_values(ascending=False)
district_status
```

```python
result = df.groupby("구")["거래금액(만원)"].agg(평균거래금액=("mean"), 거래건수=("count"), 중앙거래금액=("median")).sort_values(by='평균거래금액', ascending=False)
result # 구를 기준으로 그룹화한 후 개래금액을 평균(내림차순으로 정렬), 거래건수, 중앙값을 출력
```

```python
# 연도별 분석 2020년~2025년 거래금액 변화. 1.월별 분석 2.연도별 분석
monthly_research = df.groupby("계약년월")["거래금액(만원)"].agg(월별평균="mean", 월별중앙값="median", 월별거래건수="count").sort_index()
monthly_research
```

```python
df["계약년도"] = df["계약년월"] // 100
year_research = df.groupby("계약년도")["거래금액(만원)"].agg(연별평균="mean", 연별중앙값="median", 연별거래건수="count").sort_index()
year_research
```

```python
# 전용면적에 따른 거래금액 차이로 변경
# 전용면적이 클수록 아파트 거래금액이 높아질까?
```

```python
df["면적구간"] = pd.cut(
    df["전용면적(㎡)"],
    bins=[0, 40, 60, 85, 135, float("inf")],
    labels=["40㎡ 미만", "40~60㎡", "60~85㎡", "85~135㎡", "135㎡ 이상"],
    right=False
)
area_status = df.groupby("면적구간", observed=True)["거래금액(만원)"].agg(
    평균거래금액="mean",
    중앙거래금액="median",
    거래건수="count"
).reset_index() #면적구간을 인덱스에서 일반 컬럼으로 변경
area_status
```

```python
# 면적이 커질 수록 평균 거래금액이 높아졌음.
# 모든 면적구간에서 평균이 중앙값보다 높음. -> 일부의 고가 거래들이 평균을 올림.
# 60~85㎡ 구간에서의 거래 건수가 가장 높음.
```

```python
# ㎡당 거래금액 분석.
df["㎡당거래금액"] = df["거래금액(만원)"] / df["전용면적(㎡)"]
df["㎡당거래금액"] # 각 아파트 별 1㎡당 거래금액(만원)
```

```python
area_result = df.groupby("면적구간", observed=True)["㎡당거래금액"].agg(
    **{"평균㎡당거래금액": "mean", "중앙㎡당거래금액": "median", "거래건수": "count"}
)
area_result # ㎡는 키워드인자, 딕셔너리로
```

```python
#거래 건수는 60~85㎡이 가장 많은데 ㎡당거래금액은 40~60㎡, 85~135㎡	이 평균과 중앙값 모두 더 높네
# 단순 거래가 많다고 해서 거래금액이 높은 것은 아니구나.
```
